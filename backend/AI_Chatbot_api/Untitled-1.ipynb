```python
import json
from base64 import b64decode, b64encode
from cryptography.hazmat.primitives.asymmetric.padding import OAEP, MGF1
from cryptography.hazmat.primitives.asymmetric import rsa
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.primitives.serialization import load_pem_private_key
from cryptography.hazmat.backends import default_backend
from django.http import JsonResponse, HttpResponse
from django.views.decorators.csrf import csrf_exempt
from django.utils.crypto import constant_time_compare

# ✅ Webhook Token for WhatsApp Flow Health Check
WHATSAPP_VERIFY_TOKEN = "12345"

# ✅ Load the Private Key for RSA Decryption
def load_private_key(file_path, passphrase):
    """Load an encrypted RSA private key from a file using the given passphrase."""
    with open(file_path, "rb") as key_file:
        private_key = load_pem_private_key(
            key_file.read(),
            password=passphrase.encode(),  # Convert passphrase to bytes
            backend=default_backend()
        )
    return private_key

# 🔹 Private Key Path & Passphrase
private_key_path = "D:/meta_api_implement/backend/AI_Chatbot_api/pem_file/new_private.pem"
passphrase = "Nerdshub@123"

# 🔹 Load Private Key
try:
    private_key = load_private_key(private_key_path, passphrase)
    print("✅ Private Key Loaded Successfully!")
except Exception as e:
    print(f"❌ Error Loading Private Key: {e}")
    private_key = None

# ✅ WhatsApp Webhook (GET for Verification)
@csrf_exempt
def whatsapp_webhook(request):
    """Handles WhatsApp Webhook Verification (GET) & Messages (POST)."""

    if request.method == "GET":
        # Extract query parameters for verification
        mode = request.GET.get("hub.mode")
        token = request.GET.get("hub.verify_token")
        challenge = request.GET.get("hub.challenge")

        # Validate token & respond
        if mode == "subscribe" and constant_time_compare(token, WHATSAPP_VERIFY_TOKEN):
            print("✅ WhatsApp Webhook Verified Successfully!")
            return HttpResponse(challenge, content_type="text/plain")

        print("❌ WhatsApp Webhook Verification Failed!")
        return JsonResponse({"error": "Invalid token"}, status=403)

    elif request.method == "POST":
        try:
            body = json.loads(request.body)
            print("📩 Incoming WhatsApp Webhook Data:", json.dumps(body, indent=2))

            # ✅ Handle incoming messages (customize this part)
            # You can process incoming messages here, like saving them to a database
            return JsonResponse({"status": "received"}, status=200)

        except Exception as e:
            print(f"❌ Error Handling WhatsApp Webhook: {e}")
            return JsonResponse({"error": "Invalid request"}, status=400)

    return JsonResponse({"error": "Invalid method"}, status=405)

# ✅ Decrypt Incoming Requests
@csrf_exempt
def data(request):
    """Django API to handle decryption requests."""
    try:
        if private_key is None:
            raise ValueError("❌ Private Key is not loaded. Check passphrase or file path.")

        # Parse the request JSON body
        body = json.loads(request.body)

        # Read encrypted data from request
        encrypted_flow_data_b64 = body['encrypted_flow_data']
        encrypted_aes_key_b64 = body['encrypted_aes_key']
        initial_vector_b64 = body['initial_vector']

        # 🔹 Decrypt request
        decrypted_data, aes_key, iv = decrypt_request(
            encrypted_flow_data_b64, encrypted_aes_key_b64, initial_vector_b64
        )
        print("🔓 Decrypted Data:", decrypted_data)

        # 🔹 Response Data
        response = {
            "screen": "SCREEN_NAME",
            "data": {
                "some_key": "some_value"
            }
        }

        # 🔹 Encrypt response & return
        encrypted_response = encrypt_response(response, aes_key, iv)
        return HttpResponse(encrypted_response, content_type='text/plain')

    except Exception as e:
        print(f"❌ Decryption Error: {e}")
        return JsonResponse({"error": "Decryption failed"}, status=500)

def decrypt_request(encrypted_flow_data_b64, encrypted_aes_key_b64, initial_vector_b64):
    """Decrypt AES key using RSA, then decrypt flow data using AES-GCM."""
    if private_key is None:
        raise ValueError("❌ Private Key is not loaded. Check passphrase or file path.")

    # Decode Base64-encoded values
    encrypted_flow_data = b64decode(encrypted_flow_data_b64)
    encrypted_aes_key = b64decode(encrypted_aes_key_b64)
    iv = b64decode(initial_vector_b64)

    # 🔹 Decrypt AES key using RSA Private Key
    aes_key = private_key.decrypt(
        encrypted_aes_key,
        OAEP(
            mgf=MGF1(algorithm=hashes.SHA256()), 
            algorithm=hashes.SHA256(), 
            label=None
        )
    )

    # 🔹 Decrypt the Flow Data using AES-GCM
    encrypted_flow_data_body = encrypted_flow_data[:-16]  # Ciphertext
    encrypted_flow_data_tag = encrypted_flow_data[-16:]  # GCM Tag
    decryptor = Cipher(algorithms.AES(aes_key), modes.GCM(iv, encrypted_flow_data_tag)).decryptor()
    decrypted_data_bytes = decryptor.update(encrypted_flow_data_body) + decryptor.finalize()

    decrypted_data = json.loads(decrypted_data_bytes.decode("utf-8"))
    return decrypted_data, aes_key, iv

def encrypt_response(response, aes_key, iv):
    """Encrypt response using AES-GCM."""
    # 🔹 Flip the IV (XOR with 0xFF for some additional security)
    flipped_iv = bytes([b ^ 0xFF for b in iv])

    # 🔹 Encrypt the response data
    encryptor = Cipher(algorithms.AES(aes_key), modes.GCM(flipped_iv)).encryptor()
    ciphertext = encryptor.update(json.dumps(response).encode("utf-8")) + encryptor.finalize()

    # 🔹 Return the Base64-encoded encrypted response
    return b64encode(ciphertext + encryptor.tag).decode("utf-8")
```
